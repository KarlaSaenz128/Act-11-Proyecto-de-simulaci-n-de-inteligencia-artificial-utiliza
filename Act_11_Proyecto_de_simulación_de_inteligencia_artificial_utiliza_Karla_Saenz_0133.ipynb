```python
# --------------------------------------------------------
# PROGRAMA: SIMULADOR DE INFLUENCER (MENU INTELIGENTE)
# --------------------------------------------------------

# Creamos la lista que empezará vacía para guardar cada influencer
lista_influencer = []


# --- FUNCIONES ---
def registrar_influencer():
    print("\n--- 📝 REGISTRAR DATOS DEL INFLUENCER ---")

    # Se crea el diccionario vacío para el influencer actual
    influencer = {}

    # Entrada de datos (Key y Values)
    influencer["Nombre"] = input("👉 Nombre del influencer: ")
    influencer["Nicho de mercado"] = input("👉 Nicho de mercado (ej. Gamer, Moda): ")
    influencer["Plataforma principal"] = input(
        "👉 Plataforma principal (ej. TikTok, Instagram): "
    )
    influencer["Tasa de engagement"] = input("👉 Tasa de engagement (ej. 5%): ")
    influencer["Colaboraciones de marca"] = input(
        "👉 Colaboraciones de marca (Sí/No): "
    )
    influencer["Número de seguidores"] = input("👉 Número de seguidores (ej. 500K): ")
    influencer["Frecuencia de posteo"] = input("👉 Frecuencia de posteo (ej. Diario): ")

    # Agregamos el diccionario a la lista
    lista_influencer.append(influencer)
    print(f"\n✅ ¡Influencer '{influencer['Nombre']}' guardado con éxito!")


def mostrar_influencer():
    print("\n--- 📊 TABLA DE INFLUENCER ---")

    # Evaluamos si la lista no tiene datos
    if len(lista_influencer) == 0:
        print("No hay ningún influencer registrado todavía.")
    else:
        # Cabecera de la tabla con buena distribución de espacios
        print(
            f"{'Nombre':<15} | {'Nicho':<12} | {'Plataforma':<12} | {'Engagement':<10} | {'Colaboración':<12} | {'Seguidores':<10} | {'Frecuencia':<12}"
        )
        print("-" * 95)

        # Ciclo FOR para recorrer cada diccionario de la lista
        for inf in lista_influencer:
            print(
                f"{inf['Nombre']:<15} | {inf['Nicho de mercado']:<12} | {inf['Plataforma principal']:<12} | {inf['Tasa de engagement']:<10} | {inf['Colaboraciones de marca']:<12} | {inf['Número de seguidores']:<10} | {inf['Frecuencia de posteo']:<12}"
            )


def simular_ia_influencer():
    print("\n--- 🤖 SISTEMA INTELIGENTE DE INFLUENCER ---")

    if len(lista_influencer) == 0:
        print("El sistema no tiene datos de influencer para analizar.")
    else:
        print("¿Qué tipo de perfil deseas buscar?")
        print("1. Influencer con alta frecuencia (Posteo Diario)")
        print("2. Influencer con colaboraciones activas")

        opcion_ia = input("Selecciona una opción (1 o 2): ")
        print("\n🤖 [Analizando datos del diccionario...] 🤖")

        encontrado = False

        # El ciclo recorre y las condiciones IF-ELIF deciden
        for inf in lista_influencer:
            if opcion_ia == "1":
                if inf["Frecuencia de posteo"].lower() == "diario":
                    print(
                        f"🎯 RECOMENDACIÓN: {inf['Nombre']} es ideal porque su posteo es Diario."
                    )
                    encontrado = True

            elif opcion_ia == "2":
                if inf["Colaboraciones de marca"].lower() == "sí" or inf[
                    "Colaboraciones de marca"
                ].lower() == "si":
                    print(
                        f"🎯 RECOMENDACIÓN: {inf['Nombre']} ya trabaja con marcas en su nicho ({inf['Nicho de mercado']})."
                    )
                    encontrado = True

        if not encontrado:
            print(
                "❌ No se encontró un perfil de influencer que coincida exactamente."
            )


# --- CICLO PRINCIPAL (MENU INTERACTIVO) ---
while True:
    print("\n========================================")
    print("      📱 MENU INTERACTIVO INFLUENCER     ")
    print("========================================")
    print("1. ➕ Registrar datos de Influencer")
    print("2. 📊 Mostrar listado de Influencer (Tabla)")
    print("3. 🤖 Analizar con IA de Influencer")
    print("4. ❌ Salir")

    opcion = input("\nSelecciona una opción (1-4): ")

    # Estructura IF - ELIF - ELSE para controlar el menú
    if opcion == "1":
        registrar_influencer()
    elif opcion == "2":
        mostrar_influencer()
    elif opcion == "3":
        simular_ia_influencer()
    elif opcion == "4":
        break # Add break statement here
```
